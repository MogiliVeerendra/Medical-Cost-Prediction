```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.linear_model import ElasticNet
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
file_path = "../data/processed/ml-ready-data.xlsx"
medical = pd.read_excel(file_path)
medical.head()
```

```python
medical = medical.drop(columns=["Unnamed: 0.1","Unnamed: 0"])
```

```python
x = medical.drop(columns=["expenses"])
x.head()
```

```python
y = medical["expenses"]
y.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    train_size= 0.8,
    random_state=42
)
```

```python
print("x_train:",x_train.shape)
print("x_test:",x_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
model = LinearRegression()
```

```python
print(model.coef_)
print(model.intercept_)
```

```python
model.fit(x_train,y_train)
```

```python
print("intercept:", model.intercept_)
print("cofficients:", model.coef_)
print("no.of feature after training:", model.n_features_in_)
```

```python
for feature, coefficient in zip(x_train.columns, model.coef_):
    print(feature, ":", coefficient)
print("intercept:", model.intercept_)
```

```python
y_pred = model.predict(x_test)
```

```python
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_pred)
```

```python
print("MEAN ABSOLUTE ERROR:",mae)
print("MEAN SQUARED ERROR:",mse)
print("ROOT MEAN SQUARED ERROR:",rmse)
print("R-SQUARED:",r2)
```

```python
model_error = predication[2] - y_test.iloc[7]
model_error
```

```python
results = []
results.append(
    {
        "model" : "LinearRegression",
        "MAE" : mae,
        "MSE" : mse,
        "RMSE": rmse,
        "trainR2": r2,
        "testR2": r2
        }
)

results_df = pd.DataFrame(results)
results_df
```
