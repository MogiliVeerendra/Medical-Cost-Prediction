```python
#  OUTLIERS
q1 = medical["age"].quantile(0.25)
q2 = medical["age"].quantile(0.50)
q3 = medical["age"].quantile(0.75)

IQR = q3 - q1
lower_bound = q1 - 1.5 * IQR
higher_bound = q3 + 1.5 * IQR

age_outliers = medical[(medical["age"] < lower_bound) | (medical["age"] > higher_bound)]
age_outliers
```

```python
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", IQR)
print("lower_bound:", lower_bound)
print("higher_bound:", higher_bound)
```

```python
fig, ax = plt.subplots()
fig.set()

ax.set_title("outliers of age")
plt.boxplot(
    medical["age"]
)
ax.set_ylabel("Age values")
plt.show()
```

```python
Q1 = medical["expenses"].quantile(0.25)
Q3 = medical["expenses"].quantile(0.75)

IQR1 = Q3 - Q1
lower = Q1 - 1.5 * IQR1
high = Q3 + 1.5 * IQR1

expenses_outliers = medical[(medical["expenses"] < lower) | (medical["expenses"] > high)]
expenses_outliers
```

```python
print("Q1:",Q1)
print("Q3:", Q3)
print("IQR:",IQR)
print("lower_bound:", lower)
print("higher_bound:", high)
```
