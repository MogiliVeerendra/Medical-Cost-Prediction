```python
import pandas as pd
```

```python
file_path = "../data/raw/insurance.xlsx"
medical = pd.read_excel(file_path)
medical.head()
```

```python
medical.tail()
```

```python
medical.columns.tolist()
```

```python
print(medical["age"])
print("Highest age's:", medical["age"].value_counts().idxmax())
print(medical["age"].describe())
print("Age column information: ", medical["age"].info())
print("missing values:", medical["age"].isna().sum())
```

```python
print(medical["sex"].head())
print("no.of unique values:", medical["sex"].nunique())
```

```python
print(medical["sex"].value_counts())
print("Total no.of rows:", medical["sex"].value_counts().sum())
print("Highest category: ", medical["sex"].value_counts().idxmax())
print(medical["sex"].describe())
print(medical["sex"].info())
print("no.of missing values: ", medical["sex"].isna().sum())
```

```python
print(medical["bmi"].head())
print(medical["bmi"].describe())
print("no.of missing values:", medical["bmi"].isna().sum())
print(medical["bmi"].value_counts().idxmax())
```

```python
medical[medical["children"] > 0]
medical["children"].describe()
print("No.of missing values:", medical["children"].isna().sum())
```

```python
print(medical["smoker"].value_counts())
print(medical["smoker"].describe())
print("No.of missing values: ",medical["smoker"].isna().sum())
```
